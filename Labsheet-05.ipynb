```python
# Q1. Load and display image
import cv2

img = cv2.imread('images.jpg')

cv2.imshow('Image', img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Q2. Color to grayscale
import cv2
import numpy as np

img = cv2.imread('images.jpg')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

both = np.hstack((img, cv2.cvtColor(gray, cv2.COLOR_GRAY2BGR)))

cv2.imshow('Color and Gray', both)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Q3. RGB and HSV images
import cv2

img = cv2.imread('image.jpg')

rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

cv2.imshow('RGB', cv2.cvtColor(rgb, cv2.COLOR_RGB2BGR))
cv2.imshow('HSV', hsv)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Q4. Resize image
import cv2

img = cv2.imread('image.jpg')
resize = cv2.resize(img, (200, 200))

cv2.imshow('Resized Image', resize)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Q5. Rotate image
import cv2

img = cv2.imread('image.jpg')
h, w = img.shape[:2]

matrix = cv2.getRotationMatrix2D((w//2, h//2), 45, 1)
rotate = cv2.warpAffine(img, matrix, (w, h))

cv2.imshow('Rotated Image', rotate)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Q6. Flip image
import cv2

img = cv2.imread('image.jpg')

horizontal = cv2.flip(img, 1)
vertical = cv2.flip(img, 0)
both = cv2.flip(img, -1)

cv2.imshow('Horizontal', horizontal)
cv2.imshow('Vertical', vertical)
cv2.imshow('Both', both)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Q7. Crop image
import cv2

img = cv2.imread('image.jpg')

crop = img[50:200, 50:200]

cv2.imshow('Cropped Image', crop)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Q8. Face detection
import cv2

img = cv2.imread('image.jpg')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

face = cv2.CascadeClassifier(
    cv2.data.haarcascades + 'haarcascade_frontalface_default.xml'
)

faces = face.detectMultiScale(gray, 1.1, 4)

for x, y, w, h in faces:
    cv2.rectangle(img, (x,y), (x+w,y+h), (0,255,0), 2)

cv2.imshow('Face Detection', img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Q9. Eye detection
import cv2

img = cv2.imread('image.jpg')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

eye = cv2.CascadeClassifier(
    cv2.data.haarcascades + 'haarcascade_eye.xml'
)

eyes = eye.detectMultiScale(gray, 1.1, 5)

for x, y, w, h in eyes:
    cv2.rectangle(img, (x,y), (x+w,y+h), (0,255,0), 2)

cv2.imshow('Eye Detection', img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Q10. Shape detection
import cv2

img = cv2.imread('image.jpg')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

_, th = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)
contours, _ = cv2.findContours(
    th, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE
)

for c in contours:
    p = cv2.arcLength(c, True)
    a = cv2.approxPolyDP(c, 0.04*p, True)
    n = len(a)

    if n == 3:
        name = "Triangle"
    elif n == 4:
        name = "Rectangle"
    else:
        name = "Circle"

    cv2.putText(img, name, tuple(a[0][0]),
                cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0,255,0), 2)

cv2.imshow('Shapes', img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Q11. Webcam video
import cv2

cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()
    if not ret:
        break

    cv2.imshow('Webcam', frame)

    if cv2.waitKey(1) == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```

```python
# Q12. Real-time grayscale video
import cv2

cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()
    if not ret:
        break

    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    cv2.imshow('Gray Video', gray)

    if cv2.waitKey(1) == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```

```python
# Q13. Save 100 webcam frames
import cv2
import os

cap = cv2.VideoCapture(0)
os.makedirs('frames', exist_ok=True)

for i in range(100):
    ret, frame = cap.read()
    if not ret:
        break

    cv2.imwrite(f'frames/frame{i}.jpg', frame)

cap.release()
print("Frames saved successfully")
```
