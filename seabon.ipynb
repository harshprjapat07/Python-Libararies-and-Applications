```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "Name": ['Nasir', 'Rahul', 'Aman', 'Priya', 'Sara', 'Karan'],
    "Age": [20, 21, 19, 20, 22, 21],
    "Marks": [80, 90, 70, 95, 85, 75],
    "Gender": ["M", "M", "M", "F", "F", "M"],
    "Dept": ['BCA', 'MCA', 'BCA', 'BCA', 'BCA', 'MCA'],
    "Hours":[3, 5, 2, 6, 4, 3],
    "Salary":[25000, 40000, 22000, 45000, 35000, 20000]
}

df1 = pd.DataFrame(data)
sns.lineplot(x="Name", y="Marks", data=df1, hue = 'Gender')

plt.show()
```

```python
print(df1)
sns.countplot(data=df1, x = "Dept", hue = "Gender")
plt.show()
```

```python
sns.histplot(data=df1, x = 'Marks', bins = 5, kde = True)
plt.title("Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()
```

```python
print(df1)
sns.barplot(x='Name', y='Marks', data=df1)
```

```python
sns.barplot(x='Dept', y='Salary', data=df1, estimator="sum")
```

```python
sns.boxplot(x="Name", y="Marks", data=df1)
```

```python
sns.scatterplot(x="Name", y="Marks", data=df1)
```

```python
corr = df1.select_dtypes(include="number").corr()
sns.heatmap(corr, annot=True, cmap="coolwarm")
```
