```python
import matplotlib.pyplot as plt
x = ['a', 'b', 'c', 'd', 'e']
y = [1000, 2000, 3000, 4630, 5000]
z = [3000,4000,5000,6000,7000]
plt.plot(x, y, z, marker = "s", color = 'red', label = "Sales")
plt.plot(x, z, marker = "^", color = 'blue', label = "Revenue")
plt.xlabel('Days')
plt.ylabel('Income')
plt.title('Sales Chart')
plt.legend()
plt.grid()
plt.show()
```

```python
x = [1,2,3,4,5,6,7,8,9]
plt.hist(x, bins=5, edgecolor = "black")
plt.show()
```

```python
study_hrs = [1,2,3,4,5,6]
marks = [35, 40, 55, 60, 75, 90]
plt.scatter(study_hrs, marks)
plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
```

```python
labels = ['Food', 'Rent', 'Travel', 'Other']
expenses = [30, 40, 15, 15]
plt.pie(expenses, labels = labels, autopct = "%1.1f%%", startangle = 90)
plt.title("Monthly Expenses")
plt.show()
```
